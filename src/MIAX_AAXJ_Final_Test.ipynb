# Validación final AAXJ

Este notebook descarga una serie externa del ETF **AAXJ** y calcula para ella el mismo conjunto de features usado en la Parte 1, con el único objetivo de alimentar la validación final de la Parte 2.

La fuente utilizada es la página oficial de iShares para AAXJ, de la que se extrae la serie `performanceData` publicada en la sección *Growth of Hypothetical $10,000*.

La ventana temporal se restringe al periodo de la práctica: **2016-01-05** a **2021-07-16**.

```python
from pathlib import Path
import csv
import sys

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'src' else Path.cwd().resolve()
SRC_DIR = PROJECT_ROOT / 'src'
DATASET_DIR = PROJECT_ROOT / 'dataset'

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from aaxj_feature_builder import build_aaxj_features

OUTPUT_PATH = DATASET_DIR / 'aaxj_features_final_test.csv'
OUTPUT_PATH
```

## Generación del fichero final

Esta celda descarga la serie externa, la cruza con los factores Asia ex Japan ya existentes en `/dataset` y guarda el resultado final en `aaxj_features_final_test.csv`.

```python
feature_row = build_aaxj_features(output_path=OUTPUT_PATH)
print('Archivo generado:', OUTPUT_PATH)
for key in ['n_obs', 'start_date', 'end_date', 'ann_return', 'ann_vol', 'sharpe', 'sortino', 'max_drawdown', 'beta_mkt', 'beta_smb', 'beta_hml', 'beta_wml']:
    print(f'{key}: {feature_row[key]}')
```

La validación de la Parte 2 necesita que el archivo exista y contenga, al menos, las columnas de `clustering_features`. Aquí hacemos una lectura rápida del resultado.
